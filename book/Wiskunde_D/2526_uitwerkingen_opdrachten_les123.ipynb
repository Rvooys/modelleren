# Opdracht 1a
print  de eerste 30 ONEVEN getallen, startend met waarde 1.

Dit kan eenvoudig met een for loop, of als je les 2 hebt gehad, met een snelle lijst. Hieronder een aantal uitwerkingen, bekijk welke het beste voor jou past in je begrip.

```python
for i in range(1,31):
    print(2*i-1)
```

Dit is wellicht de makkelijkste, aangezien we er maar 30 hoeven te hebben en elk getal van de vorm 2i-1 een oneven getal is. Check dat je begint met 1: 2x1-1 = 2-1 = 1 dus de eerste waarde is inderdaad 1. Je wilt er 30, dus de range moet TOT (niet tot en met) 31 lopen.  Hieronder een aantal andere manieren.

```python
for i in range(1,60,2):
    print(i)

#we maken nu gebruik van de optie van range om bij 1 te beginnen, en steeds stapjes van 2 te doen, tot je bij 59 aangekomen ben (60 wordt niet meegenomen)
```

```python
#Deze manier maakt gebruik van les 2 modulo rekenen... een oneven getal is niet deelbaar door 2, dus deling door 2 geeft rest 1. 
#Je moet dan wel weten dat 1 tot 59 de eerste 30 oneven getallen geeft.
for i in range(1,60):
    print(i)

```

```python
#Stel je weet niet zeker wat de 30e oneven getal is, dan kan je ook een teller toevoegen met een while:
teller = 0
getal = 0
while teller <30:
    
    if getal%2==1:
        print(getal)
        teller = teller + 1
    getal=getal+1
```

```python
onevengetal = 123   #dit is het eerste oneven getal waar we mee willen starten
teller = 1  # we hebben er al eentje kunnen tellen, namelijk 123

print(onevengetal)  # we printen alvast het eerste oneven getal
while teller < 30:   #zolang we nog minder dan 30 oneven getallen hebben
    onevengetal = onevengetal + 2 #vervang de variabele door zichzelf 2 op te hogen (dat is dan ook oneven, dus nu 125)
    print(onevengetal) #print het nieuwe oneven getal (de eerste keer is dat dus nu 125)
    teller = teller + 1 #verhoog nu de teller... als deze NU 30 geworden is, stopt het programma hier.
```
